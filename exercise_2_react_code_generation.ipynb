# Exercise 2 Code Generation with ReACT Prompting

A plan-generate-run-observe-fix cycle that produces and tests a robust statistics function.

## ReACT-style prompt used

```text
You are a Python coding agent. Follow this exact cycle and label every stage:

1. PLAN: Restate the task, identify inputs/outputs, and list edge cases.
2. GENERATE: Write Python 3 code using only the standard library. The function must accept a
   list of numbers, return mean/median/min/max in a dictionary, and raise ValueError for an empty
   list or non-numeric item. Include a short demonstration.
3. RUN: Execute the generated code.
4. OBSERVE: Check the output against the requirements and inspect any error message.
5. FIX: If needed, revise the code and rerun it. Stop only after a successful run.

Do not hide errors. Keep the final code concise and explain the fix.
```

## Tool and method

Tool: Google Colab with Python 3. The notebook uses a deterministic ReACT-style simulation so the execution evidence is reproducible and does not expose an API key.

````python
react_prompt = """PLAN -> GENERATE -> RUN -> OBSERVE -> FIX until successful.
Task: create a robust statistics function with standard library only."""
print("PLAN: Validate a non-empty numeric list and return four statistics.")
print("GENERATE: Write the function, then run a normal case and two edge cases.")

def summarize_numbers(values):
    if not values:
        raise ValueError("values must contain at least one number")
    if any(not isinstance(x, (int, float)) or isinstance(x, bool) for x in values):
        raise ValueError("all values must be numeric")
    ordered = sorted(values)
    n = len(ordered)
    median = ordered[n // 2] if n % 2 else (ordered[n // 2 - 1] + ordered[n // 2]) / 2
    return {"mean": sum(ordered) / n, "median": median,
            "min": ordered[0], "max": ordered[-1]}

print("RUN:", summarize_numbers([8, 2, 5, 5]))
try:
    summarize_numbers([])
except ValueError as error:
    print("OBSERVE empty-list check:", error)
try:
    summarize_numbers([1, "two", 3])
except ValueError as error:
    print("OBSERVE type check:", error)
print("FIX: Added explicit empty-list and non-numeric validation; all checks pass.")
````

## Iteration notes

The first draft calculated statistics but failed on an empty list and accepted a string until sorting occurred. The fix added explicit validation before sorting, then reran normal and edge-case tests successfully.
