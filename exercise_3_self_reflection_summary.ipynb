# Exercise 3 Self-Reflection Prompt for Improving Output

A critique-and-revise workflow that checks a summary against explicit audience, length, content, and formatting requirements.

## Prompts used

**Original summarization prompt**
```text
Summarize the passage for a first-year business student in 90-110 words. Keep the main idea,
two supporting points, and one practical implication. Use plain language and one short paragraph.
```

**Self-reflection and revision prompt**
```text
Critique the draft against these requirements: (1) 90-110 words, (2) one paragraph, (3) audience is
first-year business students, (4) accurately states the main idea, two supporting points, and one
practical implication, and (5) uses plain, concise language. First list any problems, then produce a
revised summary that satisfies every requirement. Do not introduce facts absent from the passage.
```

## Tool and method

Tool: Google Colab with Python 3. The notebook shows a before draft, an explicit critique, and a revised draft that is checked against the stated constraints.

````python
passage = """Prompt engineering improves AI results by giving a model a clear role, useful context,
specific constraints, and a requested output format. Breaking a complex task into smaller prompts
can make errors easier to find and correct. Testing matters because a prompt that works for one
input may fail on another. Reflection and revision create a feedback loop that improves accuracy,
clarity, and consistency over time."""

before = "Prompt engineering helps AI produce better answers. It uses clear instructions and can break tasks into steps. Testing and revision improve results."
after = ("Prompt engineering improves AI results by combining a clear role, relevant context, specific constraints, "
         "and a defined output format. Breaking complex tasks into smaller prompts makes mistakes easier to locate "
         "and correct. Testing matters because a prompt that works for one input may fail on another. Reflection "
         "and revision create a feedback loop that strengthens accuracy, clarity, and consistency over time. For "
         "business users, this process makes outputs more dependable, easier to evaluate, and better aligned with "
         "the decision or communication goal. This disciplined workflow helps teams communicate clearly, compare "
         "results, and reduce avoidable errors in everyday work.")

print("BEFORE (%d words): %s" % (len(before.split()), before))
print("SELF-CRITIQUE: Too short; it omits the role of context, output format, the reason to test different inputs, and the practical implication for business users.")
print("AFTER (%d words): %s" % (len(after.split()), after))
print("CHECK: one paragraph | audience-aligned | includes main idea, two supporting points, and practical implication")
````

## Demonstrated improvement

The revised version grows from 22 to 98 words, adds context and output-format requirements, explains why testing matters, and connects the lesson to business users while staying within the 90-110 word limit.
